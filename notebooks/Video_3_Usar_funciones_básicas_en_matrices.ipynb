```python
import numpy as np
```

```python
# Generamos 20 números consecutivos y los reorganizamos en 4 filas por 5 columnas
m = np.arange(20).reshape(4,5)

print("\n--- Matriz creada con arange y reshape (4x5) ---")
print("Forma (.shape):", m.shape)
print("Dimensiones (.ndim):", m.ndim)
print("Tamaño total (.size):", m.size)
print("Matriz resultante:\n", m)
```

```python
# Creamos una matriz de 3 filas x 4 columnas rellena con ceros
m = np.zeros((3,4))

print("\n--- Matriz de ceros (3x4) ---")
print("Cantidad total de elementos (.size):", m.size)
print("Matriz de ceros:\n", m)
```

```python
# Generamos un arreglo unidimensional con 10 elementos (0 al 9)
m = np.arange(10)

print("\n--- Arreglo unidimensional (arange 10) ---")
print("Cantidad de elementos (.size):", m.size)
print("Arreglo resultante:\n", m)
```

```python
# Generamos 25 valores equiespaciados entre 99 y 88
m = np.linspace(99,88,25)

print("\n--- Secuencia generada con linspace ---")
print("Cantidad de elementos (.size):", m.size)
print("Arreglo generado:\n", m)
```

```python
# Guardamos la primera matriz de 24 elementos (2 bloques de 3 filas x 4 columnas)
m1 = np.arange(24).reshape(2,3,4)

# Guardamos la segunda matriz de 27 elementos (3 bloques de 3 filas x 3 columnas)
m2 = np.arange(27).reshape(3,3,3)

# Impresión de la primera matriz (24 elementos)
print("\n--- Matriz Tridimensional 3D (2x3x4) ---")
print("Cantidad total de elementos (.size):", m1.size)
print("Matriz resultante:\n", m1)

# Impresión de la segunda matriz (27 elementos)
print("\n--- Matriz Tridimensional 3D (3x3x3) ---")
print("Cantidad total de elementos (.size):", m2.size)
print("Matriz resultante:\n", m2)
```
