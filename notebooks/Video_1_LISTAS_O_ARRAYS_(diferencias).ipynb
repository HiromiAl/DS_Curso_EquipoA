DIFERENCIAS ENTRE LISTAS Y ARRAYS EN PYTHON Y NUMPY


```python
import numpy as np
```

### 1. Definición y Homogeneidad de Datos

```python

# Una lista de Python permite combinar enteros, textos y flotantes.
lista_mixta = [10, "Ciencia de Datos", 3.14, True]
print("=== 1. Lista nativa con datos heterogéneos ===")
print("Lista mixta:", lista_mixta)

# Un array de NumPy exige datos homogéneos (mismo tipo de dato).
# Si mezclamos tipos, NumPy convertirá automáticamente todo al tipo más general (ej. texto).
array_numerico = np.array([10, 20, 30])
print("\n=== Array de NumPy (datos homogéneos) ===")
print("Array de enteros:", array_numerico)
print("Tipo de dato (.dtype):", array_numerico.dtype)
```

### 2. Comportamiento del operador suma (+)

```python
lista_a = [1, 2, 3]
lista_b = [4, 5, 6]

array_a = np.array(lista_a)
array_b = np.array(lista_b)

# En listas nativas, '+' realiza una CONCATENACIÓN de elementos:
print("\n=== 2. Comportamiento del operador (+) ===")
print("Suma de listas (Concatenación):", lista_a + lista_b)

# En NumPy, '+' realiza una SUMA MATEMÁTICA elemento a elemento (vectorizada):
print("Suma de arrays NumPy (Suma vectorial):", array_a + array_b)
```

### 3. Operaciones Aritméticas Directas (Escalares y Multiplicación)

```python
# Sumar un escalar directamente a cada elemento de la estructura
print("\n=== 3. Operaciones Aritméticas Directas ===")
print("Array original + 10:", array_a + 10)
print("Multiplicación elemento a elemento (array_a * array_b):", array_a * array_b)
```
